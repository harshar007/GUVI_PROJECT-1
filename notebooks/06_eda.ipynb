# Step 6: Exploratory Data Analysis (EDA)
Univariate, Bivariate, Multivariate, Trend, Correlation, and Distribution analysis.


```python
import os
import pandas as pd
import plotly.express as px
from dotenv import load_dotenv
from sqlalchemy import create_engine

# 1. Load environment variables
load_dotenv()

MYSQL_HOST = os.getenv("MYSQL_HOST", "localhost")
MYSQL_PORT = os.getenv("MYSQL_PORT", "3306")
MYSQL_USER = os.getenv("MYSQL_USER", "root")
MYSQL_PASSWORD = os.getenv("MYSQL_PASSWORD", "root")
MYSQL_DB = os.getenv("MYSQL_DB", "cart2insights_db")

# 2. Connect to MySQL Database
mysql_uri = f"mysql+pymysql://{MYSQL_USER}:{MYSQL_PASSWORD}@{MYSQL_HOST}:{MYSQL_PORT}/{MYSQL_DB}"
engine = create_engine(mysql_uri)
print(f"Connected to MySQL database: {MYSQL_DB} on {MYSQL_HOST}:{MYSQL_PORT}")

# 3. Fetch data & display Plotly chart
monthly = pd.read_sql('''
    SELECT DATE_FORMAT(order_purchase_timestamp, '%%Y-%%m') AS month, COUNT(order_id) AS orders_count
    FROM orders WHERE order_purchase_timestamp IS NOT NULL
    GROUP BY month ORDER BY month
''', engine)

fig = px.line(monthly, x='month', y='orders_count', title="Monthly Order Volume Trend")
fig.show()

```
