# Step 2: Data Quality Analysis
Identifies missing values, duplicate records, inconsistent data types, potential outliers, and primary key uniqueness.


```python
import os
import sys
import os

def resolve_path(rel_path):
    curr = os.path.abspath(os.getcwd())
    while curr and os.path.dirname(curr) != curr:
        candidate = os.path.join(curr, rel_path)
        if os.path.exists(candidate):
            return os.path.abspath(candidate)
        curr = os.path.dirname(curr)
    return os.path.abspath(rel_path)
import pandas as pd

raw_dir = resolve_path("data/raw")
files = {
    "customers": "olist_customers_dataset.csv",
    "orders": "olist_orders_dataset.csv",
    "order_items": "olist_order_items_dataset.csv",
    "order_payments": "olist_order_payments_dataset.csv",
    "order_reviews": "olist_order_reviews_dataset.csv",
    "products": "olist_products_dataset.csv",
    "sellers": "olist_sellers_dataset.csv",
    "category_translation": "product_category_name_translation.csv",
    "geolocation": "olist_geolocation_dataset.csv"
}

for name, fname in files.items():
    file_path = os.path.join(raw_dir, fname)
    print(f"Checking {name} ({fname})...", flush=True)
    df = pd.read_csv(file_path)
    missing = df.isnull().sum().to_dict()
    duplicates = len(df) - len(df.drop_duplicates())
    print(f"=== {name.upper()} ===", flush=True)
    print(f"Total Rows: {len(df)}", flush=True)
    print(f"Duplicates: {duplicates}", flush=True)
    print(f"Missing Values: {missing}\n", flush=True)

```
