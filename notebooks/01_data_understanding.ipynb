# Step 1: Data Understanding & Exploration
**Cart2Insights: Decoding E-Commerce Performance**
This notebook loads all 9 raw datasets, inspects data schemas, shapes, column names, data types, and primary key relationships.


```python
import os
import sys
import os

def resolve_path(rel_path):
    curr = os.path.abspath(os.getcwd())
    while curr and os.path.dirname(curr) != curr:
        candidate = os.path.join(curr, rel_path)
        if os.path.exists(candidate):
            return os.path.abspath(candidate)
        curr = os.path.dirname(curr)
    return os.path.abspath(rel_path)
import pandas as pd

raw_dir = resolve_path("data/raw")
files = [
    "olist_customers_dataset.csv",
    "olist_orders_dataset.csv",
    "olist_order_items_dataset.csv",
    "olist_order_payments_dataset.csv",
    "olist_order_reviews_dataset.csv",
    "olist_products_dataset.csv",
    "olist_sellers_dataset.csv",
    "product_category_name_translation.csv",
    "olist_geolocation_dataset.csv"
]

for f in files:
    name = f.replace("olist_", "").replace("_dataset.csv", "").replace(".csv", "")
    df = pd.read_csv(os.path.join(raw_dir, f))
    print(f"=== DATASET: {name.upper()} ===", flush=True)
    print(f"Shape: {df.shape}", flush=True)
    print(f"Columns: {list(df.columns)}", flush=True)
    print(df.head(2), flush=True)
    print("\n", flush=True)

```
